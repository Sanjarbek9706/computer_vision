# Matplotlib grafik chizish, NumPy esa sonli massivlar yaratish uchun chaqiriladi.

```python
import matplotlib.pyplot as plt
import numpy as np
```

# Basic Line Plot — Oddiy chiziqli grafik
x va y qiymatlaridagi nuqtalar chiziq bilan ulanadi; grafik sarlavhasi va o'qlar nomlanadi.





```python
x = np.array([1, 2, 3, 4])
y = np.array([10, 20, 25, 30])

plt.plot(x,y)
plt.title("Basix Line Plot")
plt.xlabel("X values")
plt.ylabel("Y Values")
plt.show()
```

# Line Styles and Colors — Chiziq uslubi va ranglari
Chiziqni yashil va nuqtali qilib, ma'lumot nuqtalariga doira markerlari qo'shiladi.

```python
x = np.array([1, 2 , 3 ,4])
y = np. array([3, 8, 1, 10])

plt.plot(y, linestyle= 'dotted', color='green', marker='o')
plt.title("Styled Line Example")
plt.show()
```

# Multiple Lines in One Graph with Numpy and Matplotlib — Bitta grafikda bir nechta chiziq
Ikki ma'lumot qatori turli rang va markerlarda chizilib, legenda bilan ajratiladi.


```python
x = np.array([1, 2, 3, 4])
y1 = np.array([2, 4, 6, 8])
y2 = np.array([1, 3, 5, 7])

plt.plot(x, y1, label='Line 1', color='red', marker='s')
plt.plot(x, y2, label='Line 2', color='blue', linestyle='--', marker='o')
plt.title('Multiple Lines Example')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.legend()
plt.show()
```

# Bar Chart with Numpy and Maplotlib — NumPy va Matplotlib bilan ustunli diagramma
 Kategoriyalar qiymatlari balandligi har xil ustunlar orqali taqqoslanadi.

```python
x = np.array(['A', 'B', 'C', 'D'])
y = np.array([5, 9, 1, 7])

plt.bar(x, y, color = 'orange')
plt.title('Bar Chart Example')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.show()
```

# Scatter Plot with Numpy and Matplotlib — Nuqtali diagramma
Har bir x va y juftligi alohida nuqta sifatida ko'rsatiladi.

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 4, 6, 3, 7])

plt.scatter(x, y, color = 'purple', s=100)
plt.title('Scatter Plot Example')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

# Histogram with Numpy and Matplotlib — Gistogramma
Sonlar oraliqlarga ajratilib, har oraliqdagi qiymatlar soni ko'rsatiladi.

```python
data = np.array([3, 4, 5, 6, 8, 8, 8, 9, 10])

plt.hist(data, bins=5, color='skyblue', edgecolor='black')
plt.title("Histogram Example")
plt.xlabel('Values')
plt.ylabel('Count')
plt.show()
```

# Subplots (Multiple Graphs Together) — Bir oynada bir nechta grafik
Chapda chiziqli, o'ngda ustunli grafik bir qatorga joylanadi.

```python
x = np.array([1, 2, 3, 4])
y1 = np.array([2, 4, 6, 8])
y2 = np.array([3, 6, 9, 12])

plt. figure(figsize=(8, 4)) # width(wide) | 4 => height(tall)

plt.subplot(1, 2, 1)
plt.plot(x, y1, color ='red')
plt.title('Plot 1')

plt.subplot(1, 2, 2)
plt.bar(x, y2, color='blue')
plt.title('Plot 2')

plt.tight_layout()
plt.show()
```

# Save Figure Example — Grafikni faylga saqlash
Chizilgan grafik my_plot.png nomi bilan saqlanadi.

```python
x = np.array([1, 2, 3, 4])
y = np.array([4, 8, 6, 10])

plt.plot(x, y, color='green')
plt.title('Save Figure Example')
plt.savefig('my_plot.png') #Saves in current folder
plt.show()
```

# plot_wireframe(X, Y, Z)

```python
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import axes3d

plt.style.use('_mpl-gallery')

# Make data
X, Y, Z = axes3d.get_test_data(0.05)

# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.plot_wireframe(X, Y, Z, rstride=10, cstride=10)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```

# Matplotlib with Numpy Arrays — NumPy massivlari bilan grafik
Ikki ma'lumot qatori chizilib, legenda va koordinata to'ri qo'shiladi.

```python
from matplotlib.lines import lineStyles
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4])
y1 = np.array([10, 15, 20, 25])
y2 = np.array([8, 12 ,18, 22])

plt.plot(x, y1, label='Dataset A', color='red', marker='o')
plt.plot(x, y2, label='Dataset B', color='blue', marker='s', linestyle='--')
plt.title("Matplotlib with Numpy Arrays")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.legend()
plt.grid(True)
plt.show()
```
