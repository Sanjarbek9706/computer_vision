```python
# PyTorch va matplotlib'ni chaqiramiz
import torch                        # Asosiy PyTorch kutubxonasi
from torch import nn                # nn — neyron tarmoq qurish uchun tayyor "g'ishtchalar"
import matplotlib.pyplot as plt     # Grafik chizish uchun kutubxona

# PyTorch versiyasini tekshiramiz
torch.__version__                   # Versiyani ko'rsatadi → '2.11.0+cu130'
```

```python
# Qurilmani tanlaymiz
device = "cuda" if torch.cuda.is_available() else "cpu"   # GPU bo'lsa "cuda", bo'lmasa "cpu"
print(f"Using device: {device}")                         # Qaysi qurilma tanlanganini chiqaradi → cpu
```

```python
# Og'irlik va bias (biz bilgan "to'g'ri javob")
weight = 0.7                        # Og'irlik — chiziqning qiyaligi
bias = 0.3                          # Bias — chiziqning boshlanish nuqtasi

# Sonlar oralig'i
start = 0                           # Boshlanish: 0
end = 1                             # Tugash: 1 (o'zi kirmaydi)
step = 0.02                         # Qadam: 0.02 dan oshib boradi

# X va y ni yaratamiz (kirish va javob)
X = torch.arange(start, end, step).unsqueeze(dim=1)  # 0, 0.02, 0.04 ... 0.98 → 50 ta son, keyin ustun shakliga keltiriladi [50, 1]
y = weight * X + bias               # Formula: y = 0.7 * X + 0.3 → har bir X uchun javob

print("X:", X)

print("\n y:", y)


# Ma'lumotni bo'lamiz
train_split = int(0.8 * len(X))     # 50 ning 80% i → 40 . 100 data => 80 train | 20 test
X_train, y_train = X[:train_split], y[:train_split]   # Birinchi 40 tasi → o'qitish uchun
X_test, y_test = X[train_split:], y[train_split:]     # Oxirgi 10 tasi → tekshirish uchun

len(X_train), len(y_train), len(X_test), len(y_test)  # Har birining sonini ko'rsatadi → (40, 40, 10, 10)
```

```python
def plot_predictions(train_data=X_train,      # Funksiya yaratamiz; o'qitish ma'lumoti (standart: X_train)
                     train_labels=y_train,    # O'qitish javoblari (standart: y_train)
                     test_data=X_test,        # Tekshirish ma'lumoti (standart: X_test)
                     test_labels=y_test,      # Tekshirish javoblari (standart: y_test)
                     predictions=None):       # Model bashoratlari (boshida yo'q → None)
  plt.figure(figsize=(10, 7))                 # Grafik oynasini ochamiz, o'lchami 10x7

  #Plot training data in blue
  plt.scatter(train_data, train_labels, c='b', s=4, label='Training data')   # O'qitish nuqtalarini ko'k rangda chizamiz

  # Plot test data in green
  plt.scatter(test_data, test_labels, c='g', s=4, label='Testing data')      # Tekshirish nuqtalarini yashil rangda chizamiz

  if predictions is not None:                 # Agar bashorat berilgan bo'lsa...
    plt.scatter(test_data, predictions, c='r', s=4, label='Predictions')     # ...uni qizil rangda chizamiz

  plt.legend(prop={'size': 14})               # Rang izohlarini (legend) ko'rsatamiz, shrift o'lchami 14


plot_predictions()                            # Funksiyani chaqiramiz → hozircha faqat ko'k va yashil nuqtalar chiqadi
```

```python
# Subclass nn.Module to make our model
class LinearRegressionModelV2(nn.Module):        # O'z modelimizni yaratamiz; nn.Module'dan meros olamiz (barcha modellar uchun asos)
    def __init__(self):                          # Konstruktor — model yaratilganda birinchi ishlaydi
        super().__init__()                       # Ota-klass (nn.Module) konstruktorini ishga tushiramiz — majburiy qator
        # Use nn.Linear() for creating the model parameters
        self.linear_layer = nn.Linear(in_features=1,   #weight and    → kirishda 1 ta son (X)
                                      out_features=1)  # Linear => y = wx+b   → chiqishda 1 ta son (y)

    # Define the forward computation (input data x flows through nn.Linear())
    def forward(self, x: torch.Tensor) -> torch.Tensor:   # Ma'lumot model ichidan qanday o'tishini belgilaydi
        return self.linear_layer(x)                       # x ni chiziqli qatlamdan o'tkazib, natijani qaytaradi

# Set the manual seed when creating the model (this isn't always needed but is used for demonstrative purposes, try commenting it out and seeing what happens)
torch.manual_seed(42)                     # Tasodifiy sonlarni "qotiramiz" → har safar bir xil boshlang'ich qiymatlar chiqadi
model_1 = LinearRegressionModelV2()       # Modelimizdan bitta nusxa (obyekt) yaratamiz
model_1, model_1.state_dict(),            # Model tuzilishini va uning parametrlarini (weight, bias) ko'rsatamiz
```

```python
# Create loss function
loss_fn = nn.L1Loss()   # Loss Function | MAE => Mean Absolute Error   → model qancha xato qilayotganini o'lchaydi

# Create optimizer
optimizer = torch.optim.SGD(params=model_1.parameters(),  # optimize newly created model's parameters   → qaysi parametrlarni yangilash kerakligini beramiz (weight, bias)
                            lr=0.01)                      # weight, bias | equal   → har safar qancha o'zgartirish kerak (o'rganish tezligi)


# See model's untrained predictions (on test set)
model_1.eval()   #   → modelni "tekshirish rejimi"ga o'tkazamiz (o'qitish emas, faqat bashorat)

with torch.inference_mode():                  # Bu blok ichida PyTorch gradient hisoblamaydi → tezroq va kam xotira
    y_preds_early = model_1(X_test)   # backwards   → test ma'lumotini modelga berib, bashorat olamiz (backward kerak emas)

plot_predictions(predictions=y_preds_early)   # Bashoratlarni grafikda qizil nuqtalar bilan chizamiz
plt.title("Predictions before training (red = random guesses)")   # Grafikka sarlavha qo'yamiz
```

```python

# Set the number of epochs
epochs = 1000                         # Model ma'lumotni 1000 marta to'liq ko'rib chiqadi (1 epoch = 1 aylanish)

# Put data on the available device
# Without this, error will happen (not all model/data on device)
X_train = X_train.to(device)          # O'qitish kirishini tanlangan qurilmaga (cpu/cuda) ko'chiramiz
X_test = X_test.to(device)            # Test kirishini qurilmaga ko'chiramiz
y_train = y_train.to(device)          # O'qitish javoblarini qurilmaga ko'chiramiz
y_test = y_test.to(device)            # Test javoblarini qurilmaga ko'chiramiz

for epoch in range(epochs):           # 0 dan 999 gacha aylanamiz
    ### Training
    model_1.train()   # train mode is on by default after construction   → modelni o'qitish rejimiga o'tkazamiz

    # 1. Forward pass
    y_pred = model_1(X_train)         # Model o'qitish ma'lumoti uchun bashorat qiladi

    # 2. Calculate loss
    loss = loss_fn(y_pred, y_train)   # Bashorat va to'g'ri javob orasidagi xatoni hisoblaymiz

    # 3. Zero gradient optimizer
    optimizer.zero_grad()             # Oldingi aylanishdan qolgan gradientlarni 0 ga tozalaymiz

    # 4. Loss backward
    loss.backward()                   # Har bir parametr (weight, bias) xatoga qancha ta'sir qilganini hisoblaymiz

    # 5. Step the optimizer
    optimizer.step()                  # Shu hisobga qarab weight va bias ni ozgina yangilaymiz

    ### Testing
    model_1.eval()   # put the model in evaluation mode for testing (inference)   → tekshirish rejimiga o'tamiz
    # 1. Forward pass
    with torch.inference_mode():      # Gradient hisoblamaymiz (faqat tekshiryapmiz)
        test_pred = model_1(X_test)   # Test ma'lumoti uchun bashorat olamiz

        # 2. Calculate the loss
        test_loss = loss_fn(test_pred, y_test)   # Test bo'yicha xatoni hisoblaymiz

    if epoch % 100 == 0:              # Har 100-epochda (0, 100, 200, ...) ...
        print(f"Epoch: {epoch} | Train loss: {loss} | Test loss: {test_loss}")   # ...natijani chiqaramiz
```

```python
# Turn model into evaluation mode
model_1.eval()                        # Modelni tekshirish rejimiga o'tkazamiz (endi o'rganmaydi, faqat javob beradi)

# Make predictions on the test data
with torch.inference_mode():          # Gradient hisoblamaymiz → tezroq ishlaydi
    y_preds = model_1(X_test)         # O'qitilgan modeldan 10 ta test soni uchun bashorat olamiz
y_preds                               # Bashoratlarni ekranga chiqaramiz
```

```python
# Put data on the CPU and plot it
plot_predictions(predictions=y_preds.cpu())   # Bashoratlarni GPU'dan CPU'ga qaytarib, grafikda chizamiz
```
